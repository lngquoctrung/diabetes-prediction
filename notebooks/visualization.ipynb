```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.config import BRFSS_FILTERING_FILE_PATH

plt.style.use("seaborn-v0_8")
```

## **1. Load dataset**

```python
df = pd.read_csv(BRFSS_FILTERING_FILE_PATH)
df.head()
```

## **2. Visualization**

```python
df_2017 = df[df["Year"] == 2017]
df_2019 = df[df["Year"] == 2019]
df_2021 = df[df["Year"] == 2021]
```

### **2.1 Distribution of BMI**

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 7))

# BMI distribution
sns.histplot(data=df_2017, x='BMI', bins=30, ax=ax[0])
ax[0].set_title('BMI distribution in 2017')
ax[0].set_xlabel('BMI')

sns.histplot(data=df_2019, x='BMI', bins=30, ax=ax[1])
ax[1].set_title('BMI distribution in 2019')
ax[1].set_xlabel('BMI')

sns.histplot(data=df_2021, x='BMI', bins=30, ax=ax[2])
ax[2].set_title('BMI distribution in 2021')
ax[2].set_xlabel('BMI')

plt.tight_layout()
fig.suptitle("Distribution of BMI in 2017, 2019, 2021", fontsize=15)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### **2.2 Age distribution**

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 7))

# Age distribution
sns.countplot(data=df_2017, x="Age", ax=ax[0])
ax[0].set_title('Age distribution in 2017')
ax[0].set_xlabel('Age')

sns.countplot(data=df_2019, x="Age", ax=ax[1])
ax[1].set_title('Age distribution in 2019')
ax[1].set_xlabel('Age')

sns.countplot(data=df_2021, x="Age", ax=ax[2])
ax[2].set_title('Age distribution in 2021')
ax[2].set_xlabel('Age')

plt.tight_layout()
fig.suptitle("Distribution of Age in 2017, 2019, 2021", fontsize=15)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### **2.3 General Health distribution**

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 7))

# General Health distribution
sns.countplot(data=df_2017, x="GenHlth", ax=ax[0])
ax[0].set_title('GenHlth distribution in 2017')
ax[0].set_xlabel('Health level (1-5)')

sns.countplot(data=df_2019, x="GenHlth", ax=ax[1])
ax[1].set_title('GenHlth distribution in 2019')
ax[1].set_xlabel('Health level (1-5)')

sns.countplot(data=df_2021, x="GenHlth", ax=ax[2])
ax[2].set_title('GenHlth distribution in 2021')
ax[2].set_xlabel('Health level (1-5)')

plt.tight_layout()
fig.suptitle("Distribution of GenHlth in 2017, 2019, 2021", fontsize=15)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### **2.4 Disbetes distribution**

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 7))

# General Health distribution
sns.countplot(data=df_2017, x="Diabetes", ax=ax[0])
ax[0].set_title('Diabetes distribution in 2017')
ax[0].set_xlabel('Diabetes (0: No, 1: Pre-diabetes, 2: Diabetes)')

sns.countplot(data=df_2019, x="Diabetes", ax=ax[1])
ax[1].set_title('Diabetes distribution in 2019')
ax[1].set_xlabel('Diabetes (0: No, 1: Pre-diabetes, 2: Diabetes)')

sns.countplot(data=df_2021, x="Diabetes", ax=ax[2])
ax[2].set_title('Diabetes distribution in 2021')
ax[2].set_xlabel('Diabetes (0: No, 1: Pre-diabetes, 2: Diabetes)')

plt.tight_layout()
fig.suptitle("Distribution of Diabetes in 2017, 2019, 2021", fontsize=15)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### **2.5 Distribution of diabetes by indicators in each year**

```python
def diabetes_distribution_by_indicators_each_year(df, year):
    fig, ax = plt.subplots(2, 2, figsize=(12, 9))

    # Distribution of diabetes by BMI
    sns.histplot(data=df, x='BMI', bins=30, hue='Diabetes', palette='YlOrBr', ax=ax[0, 0])
    ax[0, 0].set_title('Distribution of diabetes by BMI')
    ax[0, 0].set_xlabel('BMI')

    # Distribution of diabetes by age
    sns.countplot(data=df, x='Age', hue='Diabetes', dodge=True, palette='YlOrBr', ax=ax[0, 1])
    ax[0, 1].set_title('Distribution of diabetes by age')
    ax[0, 1].set_xlabel('Age')

    # Distribution of diabetes by general health
    sns.countplot(data=df, x='GenHlth', hue='Diabetes', dodge=True, palette='YlOrBr', ax=ax[1, 0])
    ax[1, 0].set_title('Distribution of diabetes by general health')
    ax[1, 0].set_xlabel('Genernal health (1-5)')

    # Diabetes case distribution
    sns.countplot(data=df, x='Diabetes', ax=ax[1, 1])
    ax[1, 1].set_title('Diabetes case distribution')
    ax[1, 1].set_xlabel('Diabetes (0: No, 1: Pre-diabetes, 2: Diabetes)')

    fig.suptitle(f"Distribution of diabetes by indicators in {year}", fontsize=15)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```

```python
diabetes_distribution_by_indicators_each_year(df_2017, 2017)
```

```python
diabetes_distribution_by_indicators_each_year(df_2019, 2019)
```

```python
diabetes_distribution_by_indicators_each_year(df_2021, 2021)
```
